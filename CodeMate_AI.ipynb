```python

from groq import Groq

# Paste your Groq API key here
GROQ_API_KEY = "Enter your Groq API key"

client = Groq(
    api_key=GROQ_API_KEY
)

# -----------------------------------
# Select Programming Language
# -----------------------------------

print("💻 CodeMate AI")
print("Your Personal Programming Assistant\n")

print("Select your programming language:")
print("1. Python")
print("2. C")
print("3. C++")
print("4. Java")

choice = input("\nEnter your choice (1-4): ")

languages = {
    "1": "Python",
    "2": "C",
    "3": "C++",
    "4": "Java"
}

language = languages.get(choice)

if language is None:
    print("❌ Invalid choice. Please restart the program.")
    exit()

print(f"\n✅ Selected Language: {language}")
print("Type 'exit' to quit.\n")

# -----------------------------------
# System Prompt
# -----------------------------------

SYSTEM_PROMPT = f"""
You are CodeMate AI, a helpful programming assistant specialized in {language}.

Your job is to:

- Explain {language} programming concepts in simple language.
- Help users debug {language} code.
- Find and explain errors.
- Provide corrected {language} code.
- Explain code step by step.
- Help beginners learn {language}.
- Suggest better and more efficient solutions.
- Explain time and space complexity when appropriate.
- Help users solve programming and DSA problems using {language}.

When fixing code:

1. Identify the error.
2. Explain why the error occurs.
3. Provide the corrected code.
4. Briefly explain the correction.

Always use {language} unless the user specifically asks
to convert the code to another programming language.

Be friendly, clear and beginner-friendly.
"""

messages = [
    {
        "role": "system",
        "content": SYSTEM_PROMPT
    }
]

# -----------------------------------
# Chat Loop
# -----------------------------------

while True:

    user_input = input("You: ")

    if user_input.lower().strip() in ["exit", "quit"]:
        print("\nCodeMate: Happy coding! 👨‍💻")
        break

    messages.append({
        "role": "user",
        "content": user_input
    })

    try:

        completion = client.chat.completions.create(
            model="openai/gpt-oss-120b",
            messages=messages,
            temperature=1,
            max_completion_tokens=2048,
            top_p=1,
            reasoning_effort="medium",
            stream=True
        )

        print("\nCodeMate: ", end="")

        assistant_response = ""

        for chunk in completion:
            text = chunk.choices[0].delta.content or ""

            print(text, end="", flush=True)

            assistant_response += text

        print("\n")

        # Save AI response to conversation history
        messages.append({
            "role": "assistant",
            "content": assistant_response
        })

    except Exception as e:
        print("\n❌ Error:", e)
```
